```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime

# List of 45 Thai Tech Stocks on SET
tickers = [
    "DELTA.BK", "ADVANC.BK", "INTUCH.BK", "KCE.BK", "HANA.BK",
    "COM7.BK", "SYNEX.BK", "SIS.BK", "FORTH.BK", "SVI.BK",
    "SKY.BK", "THCOM.BK", "INET.BK", "HUMAN.BK", "NETBAY.BK",
    "ILINK.BK", "MSC.BK", "PT.BK", "MFEC.BK", "SICT.BK",
    "AIT.BK", "GUNKUL.BK", "SAMART.BK", "SAMTEL.BK", "SYMC.BK",
    "LOXLEY.BK", "TKS.BK", "FSMART.BK", "JTS.BK", "IT.BK",
    "JMART.BK", "TRUE.BK", "JAS.BK", "AKR.BK",
    "PLANET.BK", "TKC.BK", "SABUY.BK", "BBIK.BK", "INSET.BK",
    "JPARK.BK", "PROEN.BK", "SVOA.BK", "TQM.BK", "WICE.BK"
]

# Define the risk-free rate (can be updated with a more accurate value if needed)
risk_free_rate = 0.02  # Example: 2% as the risk-free rate

# Function to fetch the market capitalization
def get_market_cap(ticker):
    stock = yf.Ticker(ticker)
    market_cap = stock.info.get('marketCap', np.nan)
    return market_cap

# Function to calculate Beta
def get_beta(ticker, benchmark='^SET.BK'):
    stock = yf.Ticker(ticker)
    benchmark_data = yf.Ticker(benchmark)
    stock_hist = stock.history(period="1y")
    bench_hist = benchmark_data.history(period="1y")

    stock_returns = stock_hist['Close'].pct_change().dropna()
    bench_returns = bench_hist['Close'].pct_change().dropna()

    merged = pd.concat([stock_returns, bench_returns], axis=1).dropna()
    merged.columns = ['Stock', 'Benchmark']

    covariance = np.cov(merged['Stock'], merged['Benchmark'])[0][1]
    variance = np.var(merged['Benchmark'])

    beta = covariance / variance
    return beta

# Function to calculate Sharpe Ratio
def get_sharpe_ratio(ticker):
    stock = yf.Ticker(ticker)
    hist = stock.history(period="1y")
    daily_returns = hist['Close'].pct_change().dropna()
    avg_return = daily_returns.mean()
    std_dev = daily_returns.std()
    sharpe_ratio = (avg_return - risk_free_rate) / std_dev
    return sharpe_ratio

# Function to get fundamental metrics (P/E, P/B, Dividend Yield, ROE)
def get_fundamental_data(ticker):
    stock = yf.Ticker(ticker)
    info = stock.info
    pe_ratio = info.get('trailingPE', np.nan)
    pb_ratio = info.get('priceToBook', np.nan)
    dividend_yield = info.get('dividendYield', np.nan)
    roe = info.get('returnOnEquity', np.nan)
    return pe_ratio, pb_ratio, dividend_yield, roe

# Function to calculate exit signal (using moving averages)
def get_exit_signal(ticker):
    stock = yf.Ticker(ticker)
    hist = stock.history(period="1y")
    hist['50_MA'] = hist['Close'].rolling(window=50).mean()
    hist['200_MA'] = hist['Close'].rolling(window=200).mean()
    # Simple exit signal based on current price relative to 50-day MA
    exit_signal = 1 if hist['Close'].iloc[-1] < hist['50_MA'].iloc[-1] else 0
    return exit_signal

# Function to calculate YOY Return (as % change from the first to last close price in 1y)
def get_yoy_return(ticker):
    stock = yf.Ticker(ticker)
    hist = stock.history(period="1y")
    if len(hist) < 2:
        return np.nan
    yoy_return = (hist['Close'].iloc[-1] - hist['Close'].iloc[0]) / hist['Close'].iloc[0]
    return yoy_return * 100  # as percentage

# Collect data for all stocks
stock_data = []
market_caps = []

# Fetch market capitalization for each stock
for ticker in tickers:
    market_caps.append(get_market_cap(ticker))

# Total market cap of the sector
total_market_cap = sum(market_caps)

# Loop over tickers to collect metrics
for i, ticker in enumerate(tickers):
    print(f"Processing {ticker}...")
    beta = get_beta(ticker)
    sharpe_ratio = get_sharpe_ratio(ticker)
    pe_ratio, pb_ratio, dividend_yield, roe = get_fundamental_data(ticker)
    exit_signal = get_exit_signal(ticker)

    market_cap = market_caps[i]
    market_share = (market_cap / total_market_cap) * 100 if total_market_cap > 0 else np.nan

    stock_data.append({
        'Stock': ticker,
        'Beta': beta,
        'Sharpe Ratio': sharpe_ratio,
        'P/E Ratio': pe_ratio,
        'P/B Ratio': pb_ratio,
        'Dividend Yield': dividend_yield,
        'ROE': roe,
        'Market Cap': market_cap,
        'Market Share': market_share,
        'Return YOY': np.nan,  # placeholder; will update next
        'Exit Signal': exit_signal
    })

# Convert list to DataFrame
df = pd.DataFrame(stock_data)

# Update YOY Return column
df['Return YOY'] = df['Stock'].apply(get_yoy_return)

# Reorder DataFrame columns so the last three columns are Market Cap, Market Share, and Return YOY
cols_order = ['Stock', 'Beta', 'Sharpe Ratio', 'P/E Ratio', 'P/B Ratio',
              'Dividend Yield', 'ROE', 'Market Cap', 'Market Share', 'Return YOY']
df = df[cols_order]

# Sort the DataFrame by Market Share (descending)
df.sort_values(by='Market Share', ascending=False, inplace=True)

# Format Market Cap column as a plain integer (no commas)
df['Market Cap'] = df['Market Cap'].apply(lambda x: '{:,.0f}'.format(x) if pd.notnull(x) else '')

# (Optional) Format other columns if desired (here, we keep two decimals for float values)
float_format = {
    'Beta': '{:.2f}',
    'Sharpe Ratio': '{:.2f}',
    'P/E Ratio': '{:.2f}',
    'P/B Ratio': '{:.2f}',
    'Dividend Yield': '{:.2f}',
    'ROE': '{:.2f}',
    'Market Share': '{:.2f}',
    'Return YOY': '{:.2f}'
}

# Display the DataFrame with custom formatting
#print(df.to_string(index=False))
display(df)
# Alternatively, if using a Jupyter notebook:
# display(df.style.format(float_format))

# Plot: x-axis = Return YOY (%), y-axis = Beta
fig, ax = plt.subplots(figsize=(12, 8))
ax.scatter(df['Return YOY'].astype(float), df['Beta'].astype(float), color='blue', s=100)
for i, ticker in enumerate(df['Stock']):
    ax.annotate(ticker, (df['Return YOY'].astype(float).iloc[i], df['Beta'].astype(float).iloc[i]),
                fontsize=8, ha='right')

# Add a red dashed vertical line at Return YOY = 0
ax.axvline(x=0, color='red', linestyle='--', linewidth=2)

ax.set_xlabel('Return YOY (%)')
ax.set_ylabel('Beta')
ax.set_title('Return YOY vs Beta of Thai Tech Stocks')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = df[['Stock','Market Cap', 'Market Share']]

###############################################################################
# STEP 0: (Optional) Convert Market Cap to numeric if stored as strings with commas
###############################################################################
# If your df["Market Cap"] is a string like "832,626,098,176", remove commas and convert:
if df["Market Cap"].dtype == object:
    df["Market Cap"] = df["Market Cap"].str.replace(",", "").astype(float)

###############################################################################
# STEP 1: Original Data (df)
# df is assumed to have at least these columns:
# ['Stock', 'Market Share', 'Market Cap']
# Market Share is in percentage (e.g., 35.0 means 35%)
# Market Cap is in original currency units.
###############################################################################

# Helper function to calculate HHI
def calculate_hhi(data):
    """Sum of squared market shares."""
    return (data['Market Share'] ** 2).sum()

# Helper function to calculate CR4
def calculate_cr4(data):
    """Sum of top 4 market shares."""
    top4 = data.sort_values(by='Market Share', ascending=False).head(4)
    return top4['Market Share'].sum()

###############################################################################
# STEP 2: Scenario A - Splitting Giant Firms (Threshold = 20%)
###############################################################################

def split_giant_firms(data, threshold=20.0, conversion_factor=1):
    """
    Split any firm with Market Share > threshold into multiple new entities,
    each at or below the threshold. The new rows get a proportional share
    of the original firm's market cap (converted to krrp).

    Parameters:
      data: DataFrame with at least ['Stock', 'Market Share', 'Market Cap']
      threshold: Maximum allowed market share for one entity (e.g., 20.0)
      conversion_factor: Factor to convert Market Cap to krrp (e.g., 1e9 for billions)
    """
    scenario_data = data.copy()
    scenario_data = scenario_data.sort_values(by='Market Share', ascending=False).reset_index(drop=True)

    new_rows = []
    drop_indices = []

    for idx, row in scenario_data.iterrows():
        share = row['Market Share']
        if share > threshold:
            drop_indices.append(idx)
            original_share = share
            original_cap = row['Market Cap']

            # Calculate how many full chunks of 'threshold' we can have and the leftover share
            count_full = int(share // threshold)  # number of full 20% chunks
            leftover = share % threshold          # remaining share

            # For each full chunk, allocate proportional market cap
            for i in range(count_full):
                new_share = threshold
                new_cap = (new_share / original_share) * original_cap  # portion of market cap
                new_cap_krrp = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_Split{i+1}",
                    'Market Share': new_share,
                    'Market Cap': new_cap_krrp
                })
            # For leftover share, if any:
            if leftover > 0:
                new_share = leftover
                new_cap = (new_share / original_share) * original_cap
                new_cap_krrp = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_SplitRemainder",
                    'Market Share': new_share,
                    'Market Cap': new_cap_krrp
                })

    # Drop the original giant firms that were split
    scenario_data.drop(drop_indices, inplace=True)

    # Append the new split entities as a DataFrame
    new_df = pd.DataFrame(new_rows)

    # For firms that were not split, convert their Market Cap to krrp too:
    scenario_data['Market Cap'] = scenario_data['Market Cap'] / conversion_factor

    # Combine non-split and new split rows
    scenario_data = pd.concat([scenario_data, new_df], ignore_index=True)

    # Re-sort and return
    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

###############################################################################
# STEP 3: Scenario B - Merging Small Firms (Threshold = 1%)
###############################################################################

def merge_small_firms(data, threshold=1.0, new_name='Merged_Small_Firms'):
    """
    Merge all firms with Market Share < threshold into one combined entity.
    The new firm's market cap is the sum of the merged firms' market cap.
    """
    scenario_data = data.copy()
    small_firms = scenario_data[scenario_data['Market Share'] < threshold]

    # Sum market share and market cap for small firms
    merged_share = small_firms['Market Share'].sum()
    merged_cap = small_firms['Market Cap'].sum()  # already in krrp if coming from split function

    # Drop those small firms
    scenario_data = scenario_data[scenario_data['Market Share'] >= threshold]

    # If there's anything to merge, create a new merged firm row
    if merged_share > 0:
        new_row = pd.DataFrame({
            'Stock': [new_name],
            'Market Share': [merged_share],
            'Market Cap': [merged_cap]
        })
        scenario_data = pd.concat([scenario_data, new_row], ignore_index=True)

    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

###############################################################################
# STEP 4: Apply the scenarios
###############################################################################

# 4.1: Split giant firms over 20%
split_df = split_giant_firms(df, threshold=20.0, conversion_factor=1)

# 4.2: Optionally merge small firms under 1% (after splitting)
split_and_merged_df = merge_small_firms(split_df, threshold=1.0)

###############################################################################
# STEP 5: Calculate HHI / CR4 for each scenario
###############################################################################

# Original (for comparison, convert original Market Cap to krrp too)
df["Market Cap"] = df["Market Cap"]
hhi_original = calculate_hhi(df)
cr4_original = calculate_cr4(df)

# After Splitting Giants
hhi_split = calculate_hhi(split_df)
cr4_split = calculate_cr4(split_df)

# After Splitting + Merging Small Firms
hhi_split_merged = calculate_hhi(split_and_merged_df)
cr4_split_merged = calculate_cr4(split_and_merged_df)

print("=== Original ===")
display(df)
print(f"HHI = {hhi_original:.2f}, CR4 = {cr4_original:.2f}\n")

print("=== After Splitting Giants ===")
display(split_df)
print(f"HHI = {hhi_split:.2f}, CR4 = {cr4_split:.2f}\n")

print("=== After Splitting Giants + Merging Small Firms ===")
display(split_and_merged_df)
print(f"HHI = {hhi_split_merged:.2f}, CR4 = {cr4_split_merged:.2f}\n")


# Display HHI and CR4 results before plotting the charts
print("=== Market Concentration Metrics ===")
print(f"Original: HHI = {hhi_original:.2f}, CR4 = {cr4_original:.2f}")
print(f"After Splitting Giants: HHI = {hhi_split:.2f}, CR4 = {cr4_split:.2f}")
print(f"After Splitting + Merging Small Firms: HHI = {hhi_split_merged:.2f}, CR4 = {cr4_split_merged:.2f}\n")

###############################################################################
# STEP 6: Visualize the Final Scenario (Market Share Distribution)
###############################################################################

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(24, 10))

# -- 1) Original Market Shares --
axes[0].barh(df["Stock"], df["Market Share"], color="steelblue")
axes[0].set_title("Original Market Shares")
axes[0].invert_yaxis()
axes[0].set_xlabel("Market Share (%)")
for idx, share_val in enumerate(df["Market Share"]):
    axes[0].text(share_val + 0.5, idx, f"{share_val:.2f}%", va="center", fontsize=9)

# -- 2) After Splitting Giants --
axes[1].barh(split_df["Stock"], split_df["Market Share"], color="orange")
axes[1].set_title("After Splitting Giants > 20%")
axes[1].invert_yaxis()
axes[1].set_xlabel("Market Share (%)")
for idx, share_val in enumerate(split_df["Market Share"]):
    axes[1].text(share_val + 0.5, idx, f"{share_val:.2f}%", va="center", fontsize=9)

# -- 3) After Splitting Giants + Merging Small Firms --
axes[2].barh(split_and_merged_df["Stock"], split_and_merged_df["Market Share"], color="green")
axes[2].set_title("After Splitting + Merging Small (<1%)")
axes[2].invert_yaxis()
axes[2].set_xlabel("Market Share (%)")
for idx, share_val in enumerate(split_and_merged_df["Market Share"]):
    axes[2].text(share_val + 0.5, idx, f"{share_val:.2f}%", va="center", fontsize=9)

plt.tight_layout()
plt.show()
```

   - After each merge+split scenario, compute the HHI.
4. Keep Track of the scenario with the lowest HHI.
5. Output the “best scenario” as the AI’s recommendation for how to restructure the market via M&A (mergers & acquisitions or splits).


```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#######################################
# Step A: Retrieve and Map yfinance Data
#######################################
tickers = [
    "DELTA.BK", "ADVANC.BK", "INTUCH.BK", "KCE.BK", "HANA.BK",
    "COM7.BK", "SYNEX.BK", "SIS.BK", "FORTH.BK", "SVI.BK",
    "SKY.BK", "THCOM.BK", "INET.BK", "HUMAN.BK", "NETBAY.BK",
    "ILINK.BK", "MSC.BK", "PT.BK", "MFEC.BK", "SICT.BK",
    "AIT.BK", "GUNKUL.BK", "SAMART.BK", "SAMTEL.BK", "SYMC.BK",
    "LOXLEY.BK", "TKS.BK", "FSMART.BK", "JTS.BK", "IT.BK",
    "JMART.BK", "TRUE.BK", "JAS.BK", "AKR.BK",
    "PLANET.BK", "TKC.BK", "SABUY.BK", "BBIK.BK", "INSET.BK",
    "JPARK.BK", "PROEN.BK", "SVOA.BK", "TQM.BK", "WICE.BK"
]

# Retrieve market cap for each ticker
market_caps = []
for ticker in tickers:
    stock = yf.Ticker(ticker)
    info = stock.info
    market_cap = info.get("marketCap", np.nan)
    market_caps.append(market_cap)

# Create DataFrame and drop rows with missing market cap
df_yf = pd.DataFrame({
    "Stock": tickers,
    "Market Cap": market_caps
}).dropna(subset=["Market Cap"]).reset_index(drop=True)

# Compute market share as percentage of total market cap
total_cap = df_yf["Market Cap"].sum()
df_yf["Market Share"] = df_yf["Market Cap"] / total_cap * 100

#######################################
# Step X: Add Performance Metrics to Original yfinance Data
#######################################
def get_performance_score(ticker):
    """
    Fetch fundamental metrics from yfinance and compute a composite performance score.
    Components:
      - trailingPE: lower is better → use 1/PE,
      - ROE: higher is better → use ROE,
      - dividendYield: higher is better → use DY,
      - priceToBook (PB): lower is better → use 1/PB.
    The score is the average of available components.
    """
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
    except Exception:
        return np.nan
    pe = info.get('trailingPE', np.nan)
    pb = info.get('priceToBook', np.nan)
    roe = info.get('returnOnEquity', np.nan)
    dy = info.get('dividendYield', np.nan)
    score = 0
    count = 0
    if not np.isnan(pe) and pe > 0:
        score += 1/pe
        count += 1
    if not np.isnan(roe):
        score += roe
        count += 1
    if not np.isnan(dy):
        score += dy
        count += 1
    if not np.isnan(pb) and pb > 0:
        score += 1/pb   # Use positive 1/PB
        count += 1
    return score / count if count > 0 else np.nan

def fetch_fundamentals(ticker):
    """Fetch individual fundamental metrics from yfinance."""
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
    except Exception:
        return {"PE": np.nan, "PB": np.nan, "ROE": np.nan, "DY": np.nan}
    return {
        "PE": info.get('trailingPE', np.nan),
        "PB": info.get('priceToBook', np.nan),
        "ROE": info.get('returnOnEquity', np.nan),
        "DY": info.get('dividendYield', np.nan)
    }

# Compute performance metrics for each original ticker in df_yf
perf_scores, pe_list, pb_list, roe_list, dy_list = [], [], [], [], []
for idx, row in df_yf.iterrows():
    ticker = row['Stock']
    score = get_performance_score(ticker)
    funds = fetch_fundamentals(ticker)
    perf_scores.append(score)
    pe_list.append(funds["PE"])
    pb_list.append(funds["PB"])
    roe_list.append(funds["ROE"])
    dy_list.append(funds["DY"])

df_yf["Performance Score"] = perf_scores
df_yf["PE"] = pe_list
df_yf["PB"] = pb_list
df_yf["ROE"] = roe_list
df_yf["DY"] = dy_list

# Compute ranking and percentile for performance score
valid_mask = df_yf["Performance Score"].notna()
N_valid = valid_mask.sum()
if N_valid > 1:
    df_yf.loc[valid_mask, "Perf_Rank"] = df_yf.loc[valid_mask, "Performance Score"].rank(method="min", ascending=False)
    df_yf.loc[valid_mask, "Perf_Percentile"] = ((N_valid - df_yf.loc[valid_mask, "Perf_Rank"]) / (N_valid - 1)) * 100
else:
    df_yf["Perf_Rank"] = np.nan
    df_yf["Perf_Percentile"] = np.nan

def assign_grade(percentile):
    if percentile >= 90:
        return "A"
    elif percentile >= 80:
        return "B"
    elif percentile >= 65:
        return "C"
    elif percentile >= 40:
        return "D"
    else:
        return "F"

df_yf["Performance Grade"] = df_yf["Perf_Percentile"].apply(lambda x: assign_grade(x) if not np.isnan(x) else np.nan)

print("\n=== Original yfinance Data with Performance Metrics ===")
display(df_yf[["Stock", "Market Cap", "Market Share",
                "PE", "PB", "ROE", "DY", "Performance Score", "Perf_Percentile", "Performance Grade"]])

#######################################
# Step B: M&A Strategy Functions (with Constituents tracking)
#######################################
def calculate_hhi(data):
    return (data['Market Share'] ** 2).sum()

def calculate_cr4(data):
    top4 = data.sort_values(by='Market Share', ascending=False).head(4)
    return top4['Market Share'].sum()

def split_giant_firms(data, threshold=20.0, conversion_factor=1):
    scenario_data = data.copy().sort_values(by='Market Share', ascending=False).reset_index(drop=True)
    # Ensure a Constituents column exists; if not, assign the Stock value.
    if 'Constituents' not in scenario_data.columns:
        scenario_data['Constituents'] = scenario_data['Stock']
    new_rows = []
    drop_indices = []
    for idx, row in scenario_data.iterrows():
        share = row['Market Share']
        if share > threshold:
            drop_indices.append(idx)
            original_share = share
            original_cap = row['Market Cap']
            orig_const = row['Constituents']
            count_full = int(share // threshold)
            leftover = share % threshold
            for i in range(count_full):
                new_share = threshold
                new_cap = (new_share / original_share) * original_cap
                new_cap_conv = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_Split{i+1}",
                    'Market Share': new_share,
                    'Market Cap': new_cap_conv,
                    'Constituents': orig_const
                })
            if leftover > 0:
                new_share = leftover
                new_cap = (new_share / original_share) * original_cap
                new_cap_conv = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_SplitRemainder",
                    'Market Share': new_share,
                    'Market Cap': new_cap_conv,
                    'Constituents': orig_const
                })
    scenario_data.drop(drop_indices, inplace=True)
    scenario_data['Market Cap'] = scenario_data['Market Cap'] / conversion_factor
    new_df = pd.DataFrame(new_rows)
    scenario_data = pd.concat([scenario_data, new_df], ignore_index=True)
    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

def merge_small_firms(data, threshold=1.0, new_name='Merged_Small_Firms'):
    scenario_data = data.copy()
    if 'Constituents' not in scenario_data.columns:
        scenario_data['Constituents'] = scenario_data['Stock']
    small_firms = scenario_data[scenario_data['Market Share'] < threshold]
    merged_share = small_firms['Market Share'].sum()
    merged_cap = small_firms['Market Cap'].sum()
    merged_constituents = small_firms['Constituents'].tolist()
    scenario_data = scenario_data[scenario_data['Market Share'] >= threshold]
    if merged_share > 0:
        new_row = pd.DataFrame({
            'Stock': [new_name],
            'Market Share': [merged_share],
            'Market Cap': [merged_cap],
            'Constituents': [", ".join(merged_constituents)]
        })
        scenario_data = pd.concat([scenario_data, new_row], ignore_index=True)
    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

def find_optimal_mna_strategy(df_original, merge_thresholds=[0.5, 1.0, 2.0], split_thresholds=[15.0, 20.0, 25.0]):
    best_hhi = np.inf
    best_scenario = None
    for mth in merge_thresholds:
        for sth in split_thresholds:
            temp_df = df_original.copy()
            splitted_df = split_giant_firms(temp_df, threshold=sth, conversion_factor=1)
            final_df = merge_small_firms(splitted_df, threshold=mth, new_name='Merged_Small_Firms')
            hhi_val = calculate_hhi(final_df)
            cr4_val = calculate_cr4(final_df)
            if hhi_val < best_hhi:
                best_hhi = hhi_val
                best_scenario = {
                    'merge_threshold': mth,
                    'split_threshold': sth,
                    'HHI': hhi_val,
                    'CR4': cr4_val,
                    'df_result': final_df
                }
    return best_scenario

#######################################
# Step C: Run the Optimization (Latest Strategy)
#######################################
df_original = df_yf.copy()
best = find_optimal_mna_strategy(
    df_original,
    merge_thresholds=[0.5, 1.0, 2.0],
    split_thresholds=[15.0, 20.0, 25.0]
)

print("=== AI's Recommendation for M&A Strategy (Latest Heuristic) ===")
print(f"- Merge Threshold: {best['merge_threshold']}%")
print(f"- Split Threshold: {best['split_threshold']}%")
print(f"- Final HHI: {best['HHI']:.2f}")
print(f"- Final CR4: {best['CR4']:.2f}")
#print("\nFinal Market Structure:")
#display(best['df_result'])

#######################################
# Step D: Performance Scoring & Grading for Final Market Structure
#######################################
def get_performance_score(ticker):
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
    except Exception:
        return np.nan
    pe = info.get('trailingPE', np.nan)
    pb = info.get('priceToBook', np.nan)
    roe = info.get('returnOnEquity', np.nan)
    dy = info.get('dividendYield', np.nan)
    score = 0
    count = 0
    if not np.isnan(pe) and pe > 0:
        score += 1/pe
        count += 1
    if not np.isnan(roe):
        score += roe
        count += 1
    if not np.isnan(dy):
        score += dy
        count += 1
    if not np.isnan(pb) and pb > 0:
        score += 1/pb   # Using 1/PB now
        count += 1
    return score / count if count > 0 else np.nan

def compute_aggregate_performance(constituents):
    if isinstance(constituents, str):
        tickers_list = [x.strip() for x in constituents.split(",")]
    elif isinstance(constituents, list):
        tickers_list = constituents
    else:
        tickers_list = [constituents]
    scores = []
    for t in tickers_list:
        score = get_performance_score(t)
        if not np.isnan(score):
            scores.append(score)
    return np.mean(scores) if scores else np.nan

# Add performance metrics to final M&A result
df_final = best['df_result'].copy()
performance_scores = []
for idx, row in df_final.iterrows():
    if row['Stock'] in tickers:
        score = get_performance_score(row['Stock'])
    else:
        cons = row.get('Constituents', row['Stock'])
        score = compute_aggregate_performance(cons)
    performance_scores.append(score)
df_final['Performance Score'] = performance_scores

# Compute ranking and percentile for performance score in final DataFrame
valid_mask = df_final['Performance Score'].notna()
N_valid = valid_mask.sum()
if N_valid > 1:
    df_final.loc[valid_mask, 'Perf_Rank'] = df_final.loc[valid_mask, 'Performance Score'].rank(method='min', ascending=False)
    df_final.loc[valid_mask, 'Perf_Percentile'] = ((N_valid - df_final.loc[valid_mask, 'Perf_Rank']) / (N_valid - 1)) * 100
else:
    df_final['Perf_Rank'] = np.nan
    df_final['Perf_Percentile'] = np.nan

def assign_grade(percentile):
    if percentile >= 90:
        return "A"
    elif percentile >= 80:
        return "B"
    elif percentile >= 65:
        return "C"
    elif percentile >= 40:
        return "D"
    else:
        return "F"

df_final['Performance Grade'] = df_final['Perf_Percentile'].apply(lambda x: assign_grade(x) if not np.isnan(x) else np.nan)

# Also, fetch fundamental metrics for each row in df_final
def fetch_fundamentals(ticker):
    try:
        stock = yf.Ticker(ticker)
        info = stock.info
    except Exception:
        return {"PE": np.nan, "PB": np.nan, "ROE": np.nan, "DY": np.nan}
    return {
        "PE": info.get('trailingPE', np.nan),
        "PB": info.get('priceToBook', np.nan),
        "ROE": info.get('returnOnEquity', np.nan),
        "DY": info.get('dividendYield', np.nan)
    }

fund_PE, fund_PB, fund_ROE, fund_DY = [], [], [], []
for idx, row in df_final.iterrows():
    if row['Stock'] in tickers:
        funds = fetch_fundamentals(row['Stock'])
    else:
        cons = row.get('Constituents', row['Stock'])
        if isinstance(cons, str):
            tickers_list = [x.strip() for x in cons.split(",")]
        else:
            tickers_list = [cons]
        pe_vals, pb_vals, roe_vals, dy_vals = [], [], [], []
        for t in tickers_list:
            funds = fetch_fundamentals(t)
            if not np.isnan(funds["PE"]):
                pe_vals.append(funds["PE"])
            if not np.isnan(funds["PB"]):
                pb_vals.append(funds["PB"])
            if not np.isnan(funds["ROE"]):
                roe_vals.append(funds["ROE"])
            if not np.isnan(funds["DY"]):
                dy_vals.append(funds["DY"])
        funds = {
            "PE": np.mean(pe_vals) if pe_vals else np.nan,
            "PB": np.mean(pb_vals) if pb_vals else np.nan,
            "ROE": np.mean(roe_vals) if roe_vals else np.nan,
            "DY": np.mean(dy_vals) if dy_vals else np.nan
        }
    fund_PE.append(funds["PE"])
    fund_PB.append(funds["PB"])
    fund_ROE.append(funds["ROE"])
    fund_DY.append(funds["DY"])
df_final['PE'] = fund_PE
df_final['PB'] = fund_PB
df_final['ROE'] = fund_ROE
df_final['DY'] = fund_DY

print("\n=== Final Market Structure with Performance Metrics ===")
display(df_final[['Stock', 'Market Share', 'Market Cap', 'Constituents',
                  'PE', 'PB', 'ROE', 'DY', 'Performance Score', 'Perf_Percentile', 'Performance Grade']])

#######################################
# Step E: Plot Final Market Structure
#######################################
plt.figure(figsize=(10, 6))
plt.barh(df_final['Stock'], df_final['Market Share'], color='teal')
plt.xlabel("Market Share (%)")
plt.title("Final Market Structure after Optimal M&A Strategy")
plt.gca().invert_yaxis()
for idx, val in enumerate(df_final['Market Share']):
    plt.text(val + 0.5, idx, f"{val:.2f}%", va="center", fontsize=9)
plt.tight_layout()
plt.show()
```

### 📊 **Performance Score Calculation Methodology**

The **Performance Score** is a composite metric that combines **four fundamental indicators** into a single **average value** for better stock evaluation.

\
#### **1️⃣ Fetch Metrics**
For each stock, we retrieve the following **fundamental indicators** from **Yahoo Finance (yfinance):**
- **Trailing P/E** → Price-to-Earnings Ratio  
- **Return on Equity (ROE)**  
- **Dividend Yield (DY)**  
- **Price-to-Book (PB) Ratio**  

\

#### **2️⃣ Transform Each Metric**
To ensure that **better values contribute positively**, we transform the metrics as follows:

- #### ✅ **Trailing P/E Ratio:**
 - A **lower P/E** is considered **more attractive**.
 - **Transformation:** Use the **reciprocal** of P/E.

 $
\text{Contribution} = \frac{1}{P/E}
$

- #### ✅ **Return on Equity (ROE):**
 - A **higher ROE** is **better**.
 - **Transformation:** Use **ROE directly**.

 $
\text{Contribution} = \text{ROE}
$

- #### ✅ **Dividend Yield (DY):**
 - A **higher Dividend Yield** is **better**.
 - **Transformation:** Use **DY directly**.

 $
\text{Contribution} = \text{DY}
$

- #### ✅ **Price-to-Book (PB) Ratio:**
 - A **lower PB** is **more attractive**.
 - **Transformation:** Use the **negative reciprocal** of PB.

 $
\text{Contribution} = \frac{1}{PB}
$

\

#### **3️⃣ Combine the Contributions**
For each stock, the transformed contributions from **all available metrics** are summed:

\
  $
\text{Total Contribution} = \frac{1}{P/E} + \text{ROE} + \text{DY} + \frac{1}{PB}
$

**Note:** If a metric is **missing or invalid**, it is **excluded** from the sum.

\

#### **4️⃣ Compute the Average (Final Performance Score)**
The **Performance Score** is the **average contribution per valid metric**:

 $
\text{Performance Score} = \frac{\text{Total Contribution}}{\text{Number of Valid Metrics}}
$

\

#### **📌 Key Takeaways**
✅ The **Performance Score** adjusts fundamental ratios to ensure **better values contribute positively**.  
✅ The **reciprocal transformation** ensures that **lower P/E and PB ratios** are correctly weighted.  
✅ **Missing or invalid metrics** do not distort the score.  
✅ This approach provides a **standardized comparison** across different stocks.  

🚀 **This method helps investors quickly assess the relative strength of a stock based on key financial indicators.**  


### 📊 **Performance Percentile & Grading System**

After computing the **Performance Score** for each stock, we **rank and classify** stocks into **percentiles** and **grades** to evaluate their relative performance.

\

#### **🔹 1. Ranking Stocks**
- Stocks are **ranked in descending order** based on their **Performance Score**.
- The stock with the **highest score** gets **rank 1**, while the lowest-scoring stock gets the lowest rank.

\

#### **🔹 2. Computing Performance Percentile (Perf_Percentile)**  
The **Performance Percentile** transforms the **rank** into a **relative score** between **0 and 100**.

##### ✅ **Formula:**

 $
\text{Perf_Percentile} = \left(\frac{N - \text{Rank}}{N - 1}\right) \times 100
$
where:  
- **N** = Total number of stocks with a valid **Performance Score**  
- **Rank** = Position of the stock in the **sorted list (descending order)**  

##### ✅ **Interpretation:**
- **A higher percentile** means the stock is performing **better relative to its peers**.
- **Example Calculation:**  
  - If **N = 20** and a stock ranks **3rd**, its percentile is:
  $
  \left(\frac{20 - 3}{20 - 1}\right) \times 100 = 89.47
  $
  - This stock falls **just below the A-grade threshold**.

\

#### **🔹 3. Assigning Performance Grades**
Each stock is assigned a **letter grade** based on its **Perf_Percentile**:

| **Grade** | **Percentile Range** | **Meaning** |
|-----------|----------------------|-------------|
| **A** | ≥ 90 | **Top-tier performance** |
| **B** | ≥ 80 but **< 90** | **Above average performance** |
| **C** | ≥ 65 but **< 80** | **Moderate performance** |
| **D** | ≥ 40 but **< 65** | **Below average performance** |
| **F** | < 40 | **Poor performance** |

\

#### **📌 Key Takeaways**
✅ **Performance Percentile** allows a **relative comparison** across multiple stocks.  
✅ The **grading system provides an intuitive ranking** for investors.  
✅ **Higher grades (A & B)** indicate **stronger fundamental performance**.  
✅ **Lower grades (D & F)** suggest **weaker fundamentals** or underperformance.  

🚀 **This system helps investors quickly interpret and compare stock performance!**

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#######################################
# Step A: Retrieve and Map yfinance Data
#######################################
thai_bank_tickers = [
    "SCB.BK",     # SCB X Public Company Limited
    "KBANK.BK",   # Kasikornbank Public Company Limited
    "KTB.BK",     # Krung Thai Bank Public Company Limited
    "BBL.BK",     # Bangkok Bank Public Company Limited
    "TTB.BK",     # TMBThanachart Bank Public Company Limited
    "BAY.BK",     # Bank of Ayudhya Public Company Limited
    "TISCO.BK",   # Tisco Financial Group Public Company Limited
    "TCAP.BK",    # Thanachart Capital Public Company Limited
    "KKP.BK",     # Kiatnakin Phatra Bank Public Company Limited
    "CIMBT.BK",   # CIMB Thai Bank Public Company Limited
    "CREDIT.BK",  # Thai Credit Bank Public Company Limited
    "LHFG.BK"     # LH Financial Group Public Company Limited
]

market_caps = []
for ticker in thai_bank_tickers:
    stock = yf.Ticker(ticker)
    info = stock.info
    market_cap = info.get("marketCap", np.nan)
    market_caps.append(market_cap)

df_banks = pd.DataFrame({
    "Stock": thai_bank_tickers,
    "Market Cap": market_caps
}).dropna(subset=["Market Cap"]).reset_index(drop=True)

total_cap_orig = df_banks["Market Cap"].sum()
df_banks["Market Share (%)"] = df_banks["Market Cap"] / total_cap_orig * 100

# Format numeric columns only
df_banks_formatted = df_banks.copy()
for col in df_banks_formatted.select_dtypes(include=[np.number]).columns:
    df_banks_formatted[col] = df_banks_formatted[col].map(lambda x: f'{x:,.2f}')

print("=== Original Market Structure ===")
display(df_banks_formatted)

#######################################
# Step B: Define HHI and CR4 Calculation Functions
#######################################
def calculate_hhi(data):
    return (data["Market Share (%)"] ** 2).sum()

def calculate_cr4(data):
    top4 = data.sort_values(by="Market Share (%)", ascending=False).head(4)
    return top4["Market Share (%)"].sum()

hhi_orig = calculate_hhi(df_banks)
cr4_orig = calculate_cr4(df_banks)
print(f"\nHHI (Original) = {hhi_orig:.2f}")
print(f"CR4 (Original) = {cr4_orig:.2f}\n")

#######################################
# Step C: Simulate the Merger of KTB.BK and TTB.BK -> KTBTTB
#######################################
ktb_row = df_banks[df_banks["Stock"] == "KTB.BK"]
ttb_row = df_banks[df_banks["Stock"] == "TTB.BK"]

if not ktb_row.empty and not ttb_row.empty:
    merged_cap = ktb_row["Market Cap"].values[0] + ttb_row["Market Cap"].values[0]

    # Remove KTB.BK and TTB.BK
    df_after = df_banks[~df_banks["Stock"].isin(["KTB.BK", "TTB.BK"])].copy()

    # Append the merged entity row with new name "KTBTTB" using pd.concat
    new_row = pd.DataFrame({
        "Stock": ["KTBTTB"],
        "Market Cap": [merged_cap]
    })
    df_after = pd.concat([df_after, new_row], ignore_index=True)

    # Recompute total market cap and update market shares
    total_cap_after = df_after["Market Cap"].sum()
    df_after["Market Share (%)"] = df_after["Market Cap"] / total_cap_after * 100

    # Sort the DataFrame by Market Share (%) in descending order
    df_after.sort_values(by="Market Share (%)", ascending=False, inplace=True)
    df_after.reset_index(drop=True, inplace=True)

    hhi_after = calculate_hhi(df_after)
    cr4_after = calculate_cr4(df_after)

    # Format for printing
    df_after_formatted = df_after.copy()
    for col in df_after_formatted.select_dtypes(include=[np.number]).columns:
        df_after_formatted[col] = df_after_formatted[col].map(lambda x: f'{x:,.2f}')

    print("\n=== Merged Market Structure (KTB.BK + TTB.BK -> KTBTTB) ===")
    display(df_after_formatted)
    print(f"\nHHI (After Merge) = {hhi_after:.2f}")
    print(f"CR4 (After Merge) = {cr4_after:.2f}\n")

    #######################################
    # Step D: Plot Market Cap Before and After Merge
    #######################################
    fig, axs = plt.subplots(1, 2, figsize=(14, 6))

    axs[0].barh(df_banks["Stock"], df_banks["Market Share (%)"], color="skyblue")
    axs[0].set_title("Market Share (%) Before Merge")
    axs[0].set_xlabel("Market Share (%)")
    axs[0].invert_yaxis()

    axs[1].barh(df_after["Stock"], df_after["Market Share (%)"], color="salmon")
    axs[1].set_title("Market Share (%) After Merge")
    axs[1].set_xlabel("Market Share (%)")
    axs[1].invert_yaxis()

    plt.tight_layout()
    plt.show()
else:
    print("Either KTB.BK or TTB.BK data is missing.")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

#######################################
# Step A: Retrieve and Map yfinance Data
#######################################
tickers = [
    "SCB.BK",     # SCB X Public Company Limited
    "KBANK.BK",   # Kasikornbank Public Company Limited
    "KTB.BK",     # Krung Thai Bank Public Company Limited
    "BBL.BK",     # Bangkok Bank Public Company Limited
    "TTB.BK",     # TMBThanachart Bank Public Company Limited
    "BAY.BK",     # Bank of Ayudhya Public Company Limited
    "TISCO.BK",   # Tisco Financial Group Public Company Limited
    "TCAP.BK",    # Thanachart Capital Public Company Limited
    "KKP.BK",     # Kiatnakin Phatra Bank Public Company Limited
    "CIMBT.BK",   # CIMB Thai Bank Public Company Limited
    "CREDIT.BK",  # Thai Credit Bank Public Company Limited
    "LHFG.BK"     # LH Financial Group Public Company Limited
]

# Retrieve market cap for each ticker
market_caps = []
for ticker in tickers:
    stock = yf.Ticker(ticker)
    info = stock.info
    market_cap = info.get("marketCap", np.nan)
    market_caps.append(market_cap)

df_yf = pd.DataFrame({
    "Stock": tickers,
    "Market Cap": market_caps
}).dropna(subset=["Market Cap"]).reset_index(drop=True)

# Compute market share as percentage of total market cap
total_cap = df_yf["Market Cap"].sum()
df_yf["Market Share"] = df_yf["Market Cap"] / total_cap * 100

# Save a copy for optimization
df_original = df_yf.copy()

#######################################
# Step B: M&A Strategy Functions (with Constituents tracking)
#######################################
def calculate_hhi(data):
    return (data['Market Share'] ** 2).sum()

def calculate_cr_n(data, n):
    topn = data.sort_values(by='Market Share', ascending=False).head(n)
    return topn['Market Share'].sum()

def split_giant_firms(data, threshold=20.0, conversion_factor=1):
    scenario_data = data.copy().sort_values(by='Market Share', ascending=False).reset_index(drop=True)
    if 'Constituents' not in scenario_data.columns:
        scenario_data['Constituents'] = scenario_data['Stock']
    new_rows = []
    drop_indices = []
    for idx, row in scenario_data.iterrows():
        share = row['Market Share']
        if share > threshold:
            drop_indices.append(idx)
            original_share = share
            original_cap = row['Market Cap']
            orig_const = row['Constituents']
            count_full = int(share // threshold)
            leftover = share % threshold
            for i in range(count_full):
                new_share = threshold
                new_cap = (new_share / original_share) * original_cap
                new_cap_conv = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_Split{i+1}",
                    'Market Share': new_share,
                    'Market Cap': new_cap_conv,
                    'Constituents': orig_const
                })
            if leftover > 0:
                new_share = leftover
                new_cap = (new_share / original_share) * original_cap
                new_cap_conv = new_cap / conversion_factor
                new_rows.append({
                    'Stock': f"{row['Stock']}_SplitRemainder",
                    'Market Share': new_share,
                    'Market Cap': new_cap_conv,
                    'Constituents': orig_const
                })
    scenario_data.drop(drop_indices, inplace=True)
    scenario_data['Market Cap'] = scenario_data['Market Cap'] / conversion_factor
    new_df = pd.DataFrame(new_rows)
    scenario_data = pd.concat([scenario_data, new_df], ignore_index=True)
    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

def merge_small_firms(data, threshold=1.0, new_name='Merged_Small_Firms'):
    scenario_data = data.copy()
    if 'Constituents' not in scenario_data.columns:
        scenario_data['Constituents'] = scenario_data['Stock']
    small_firms = scenario_data[scenario_data['Market Share'] < threshold]
    merged_share = small_firms['Market Share'].sum()
    merged_cap = small_firms['Market Cap'].sum()
    merged_constituents = small_firms['Constituents'].tolist()
    scenario_data = scenario_data[scenario_data['Market Share'] >= threshold]
    if merged_share > 0:
        new_row = pd.DataFrame({
            'Stock': [new_name],
            'Market Share': [merged_share],
            'Market Cap': [merged_cap],
            'Constituents': [", ".join(merged_constituents)]
        })
        scenario_data = pd.concat([scenario_data, new_row], ignore_index=True)
    scenario_data.sort_values(by='Market Share', ascending=False, inplace=True)
    scenario_data.reset_index(drop=True, inplace=True)
    return scenario_data

# Updated optimization function with revised logic for minimize mode.
def optimize_mna_for_cr(df_original, n_big, target_CR, mode="minimize"):
    if mode == "minimize":
        # For minimize mode, fix merge threshold to 0 and set split threshold to target_CR / n_big.
        merge_thresh = 0.0
        split_thresh = target_CR / n_big
        temp_df = df_original.copy()
        splitted_df = split_giant_firms(temp_df, threshold=split_thresh, conversion_factor=1)
        final_df = merge_small_firms(splitted_df, threshold=merge_thresh, new_name='Merged_Small_Firms')
        HHI_val = calculate_hhi(final_df)
        CR_n = calculate_cr_n(final_df, n_big)
        best_scenario = {
            'merge_threshold': merge_thresh,
            'split_threshold': split_thresh,
            'HHI': HHI_val,
            'CR_n': CR_n,
            'df_result': final_df
        }
        return best_scenario
    elif mode == "maximize":
        valid_scenarios = []
        merge_range = np.arange(0.0, 20.1, 1.0)
        split_range = np.arange(1.0, 100.1, 5.0)
        for merge_thresh in merge_range:
            for split_thresh in split_range:
                temp_df = df_original.copy()
                splitted_df = split_giant_firms(temp_df, threshold=split_thresh, conversion_factor=1)
                final_df = merge_small_firms(splitted_df, threshold=merge_thresh, new_name='Merged_Small_Firms')
                # Enforce government policy: no firm may hold more than the split threshold.
                if final_df["Market Share"].max() > split_thresh:
                    continue
                HHI_val = calculate_hhi(final_df)
                CR_n = calculate_cr_n(final_df, n_big)
                if CR_n <= target_CR:
                    valid_scenarios.append({
                        'merge_threshold': merge_thresh,
                        'split_threshold': split_thresh,
                        'HHI': HHI_val,
                        'CR_n': CR_n,
                        'df_result': final_df
                    })
        if valid_scenarios:
            best_scenario = max(valid_scenarios, key=lambda x: x['HHI'])
        else:
            best_scenario = None
        return best_scenario

#######################################
# Step C: Interactive Optimization & Visualization with Custom Controls
#######################################
output = widgets.Output()

def run_custom(n_big, target_CR, mode):
    best = optimize_mna_for_cr(df_original, n_big, target_CR, mode=mode)
    with output:
        clear_output(wait=True)
        if best is None:
            display(widgets.HTML(value="<h2>No valid scenario found under current constraints.</h2>"))
            return
        card_html = f"""
        <h2>Optimal M&A Strategy Recommendation</h2>
        <p><b>Desired Big Firms (n):</b> {n_big}</p>
        <p><b>Target CR (Top {n_big} Firms):</b> {target_CR:.2f}%</p>
        <p><b>Mode:</b> {mode}</p>
        <p><b>Optimal Merge Threshold:</b> {best['merge_threshold']:.2f}%</p>
        <p><b>Optimal Split Threshold (Policy Limit):</b> {best['split_threshold']:.2f}%</p>
        <p><b>Final HHI:</b> {best['HHI']:.2f}</p>
        <p><b>Final CR (Top {n_big} Firms):</b> {best['CR_n']:.2f}%</p>
        """
        display(widgets.HTML(value=card_html))
        print("=== Final Market Structure ===")
        display(best['df_result'])
        plt.figure(figsize=(10, 6))
        plt.barh(best['df_result']['Stock'], best['df_result']['Market Share'], color='teal')
        plt.xlabel("Market Share (%)")
        plt.title("Final Market Structure after Optimal M&A Strategy")
        plt.gca().invert_yaxis()
        for idx, val in enumerate(best['df_result']['Market Share']):
            plt.text(val + 0.5, idx, f"{val:.2f}%", va="center", fontsize=9)
        plt.tight_layout()
        plt.show()

# Create interactive widgets for custom controls (without manual merge/split sliders)
n_big_slider = widgets.IntSlider(
    value=4,
    min=1,
    max=10,
    step=1,
    description='Big Firms (#)',
    continuous_update=False
)

target_CR_slider = widgets.FloatSlider(
    value=50.0,
    min=0,
    max=100,
    step=1,
    description='Target CR (%)',
    continuous_update=False
)

mode_dropdown = widgets.Dropdown(
    options=[("Minimize HHI", "minimize"), ("Maximize HHI (CR constraint)", "maximize")],
    value="minimize",
    description="Mode:"
)

optimize_button = widgets.Button(
    description="Optimize",
    button_style='success'
)

def on_optimize_button_clicked(b):
    run_custom(n_big_slider.value, target_CR_slider.value, mode_dropdown.value)

optimize_button.on_click(on_optimize_button_clicked)

controls = widgets.VBox([n_big_slider, target_CR_slider, mode_dropdown, optimize_button])
ui = widgets.HBox([controls, output])
display(ui)
```
